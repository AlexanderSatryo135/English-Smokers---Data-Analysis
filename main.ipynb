Data Visualisation // English Smokers Count by Age Group

```python
import matplotlib.pyplot as plt

young_smokers = df[(df["smoke"] == "Yes") & (df["age"].between(18, 25)) & (df["nationality"] == "English")].shape[0]
adult_smokers = df[(df["smoke"] == "Yes") & (df["age"].between(26, 40)) & (df["nationality"] == "English")].shape[0]

age_groups = ['18 - 25 Years', '26 - 40 Years']
counts = [young_smokers, adult_smokers]

plt.figure(figsize=(7, 7))

plt.pie(counts, labels=age_groups, colors=['skyblue', 'salmon'], autopct=lambda p: '{:.0f}'.format(p * sum(counts) / 100), startangle=90, explode=(0.05, 0))

plt.title('English Smokers Count by Age Group', fontsize=14, fontweight='bold', pad=20)

plt.tight_layout()
plt.show()
```

Data Visualisation // Preferred Tobacco Type by Age Group (English Smokers)

```python
import pandas as pd
import matplotlib.pyplot as plt

young_types = df[(df["smoke"] == "Yes") & (df["age"].between(18, 25)) & (df["nationality"] == "English")]["type"].value_counts()
adult_types = df[(df["smoke"] == "Yes") & (df["age"].between(26, 40)) & (df["nationality"] == "English")]["type"].value_counts()

combined_data = pd.DataFrame({
    '18 - 25 Years': young_types,
    '26 - 40 Years': adult_types
}).fillna(0)

ax = combined_data.plot(kind='bar', figsize=(10, 6), color=['skyblue', 'salmon'], width=0.7)

for container in ax.containers:
    ax.bar_label(container, fmt='%.0f', padding=3)

plt.title('Preferred Tobacco Type by Age Group (English Smokers)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Tobacco Type', fontsize=12)
plt.ylabel('Count', fontsize=12)

plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

Data Visualisation // Top 10 Smoking Patterns (Weekend / Weekday Amount) by Age Group

```python
import pandas as pd
import matplotlib.pyplot as plt

young_amt = df[(df["smoke"] == "Yes") & (df["age"].between(18, 25)) & (df["nationality"] == "English")][["amt_weekends", "amt_weekdays"]].value_counts()
adult_amt = df[(df["smoke"] == "Yes") & (df["age"].between(26, 40)) & (df["nationality"] == "English")][["amt_weekends", "amt_weekdays"]].value_counts()

young_amt.index = [f"{wnd} / {wdy}" for wnd, wdy in young_amt.index]
adult_amt.index = [f"{wnd} / {wdy}" for wnd, wdy in adult_amt.index]

combined_data = pd.DataFrame({
    '18 - 25 Years': young_amt,
    '26 - 40 Years': adult_amt
}).fillna(0)

combined_data['Total'] = combined_data['18 - 25 Years'] + combined_data['26 - 40 Years']
combined_data = combined_data.sort_values(by='Total', ascending=False).head(10).drop(columns=['Total'])

ax = combined_data.plot(kind='bar', figsize=(12, 6), color=['skyblue', 'salmon'], width=0.7)

for container in ax.containers:
    ax.bar_label(container, fmt='%.0f', padding=3)

plt.title('Top 10 Smoking Patterns (Weekend / Weekday Amount) by Age Group', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Amount (Weekends / Weekdays)', fontsize=12)
plt.ylabel('Count (People)', fontsize=12)

plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

Data Visualisation // Smokers Count by Gross Income

```python
import matplotlib.pyplot as plt

low_income_smokers = df[(df["gross_income"] == "5,200 to 10,400") & (df["smoke"] == "Yes")].shape[0]
mid_income_smokers = df[(df["gross_income"] == "10,400 to 15,600") & (df["smoke"] == "Yes")].shape[0]

income_labels = ['5,200 to 10,400', '10,400 to 15,600']
counts = [low_income_smokers, mid_income_smokers]

plt.figure(figsize=(8, 5))
bars = plt.bar(income_labels, counts, color=['mediumseagreen', 'orange'], width=0.5)

plt.bar_label(bars, fmt='%d', padding=3, fontsize=11)

plt.title('Smokers Count by Gross Income', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Gross Income Group', fontsize=12)
plt.ylabel('Number of Smokers', fontsize=12)

plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
